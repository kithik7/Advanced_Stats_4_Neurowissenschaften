# **1. Some terms and concepts**

Experimental observations are called ***data*** and deserve the utmost reverence and respect. Data are plural and the singular (***datum***) is rarely used.

The interpretation of observations typically involves statistics, a gray area that is neither quite mathematics nor quite science.

**Descriptive statistics** refers to formulas, procedures, or algorithms that summarize observations in some quantitative way (e.g., by computing a mean or a variance).

**Inferential statistics** refers to methods for analyzing the accuracy of this description. How closely does it describe the underlying (unobserved) reality? How much trust can we place in it? Can I stake my family home on it? Should it determine policy for hundreds, thousands, or millions of people?

Stastisticians of different periods and persuations have preferred different approaches -- classical, frequentist, Fisherian, Bayesian --
and the availability of massive computing power has added many further options to the menu.

We can be agnostic and adopt whatever method offers the most convincing support for our preferred conclusion. Our goal is to convince firstly ourselves and secondly our bosses and reviewers. The first should be the higher hurdle.

# More terms and concepts

There are many things that would be nice to have. Most likely, we won't get all of them:

**Validity** -- measuring what we intend (and purport) to measure. This requires a good experiment. One of my mentors used to say "Happiness is a good experiment" (Bela Julesz).

**Reliability** -- data that are internally consistent and repeatable (test-retest). This can be difficult if the object of our observations changes as a result of being observed, as is often the case with experimental animals or observers.

**Accuracy** -- how close are our measurements to the true values?  If we measure individual height with a yardstick, the results are less accurate (one, two, or three yards?) than if we use a tape measure marked in inches. Trying to measure body height in micrometers or Angstroms would be pretty meaningless.

**Precision** -- what is the spread of our measurements? How large is the difference between the smallest and the largest (are they close or dispersed)?


Note that "precision" compares measurements to each other, whereas "accuracy" compares them to ground truth.  Thus, "precision" **does not imply** "accuracy".

# Types of variables

We can call measurements or observations **variables** because they typically take different values every time we measure or observe.

These variables can be **categorical** (e.g., male or female, wild-type or mutant, ...), **binary** (e.g., zero or one, heads or tails), ordinal (first, second, third, ...), **discrete** (e.g., integers $\ldots -2, -1, 0, 1, 2 \ldots$),
or **continuous** (e.g., fractions, decimals, or real numbers).

```python
from inspect import signature
# @title Types of variables {"vertical-output":true,"display-mode":"form"}
import matplotlib.pyplot as plt
import numpy as np
from ipywidgets import interact
from scipy.stats import poisson
from scipy.stats import norm
import matplotlib.colors as mcolors
from matplotlib.colors import LinearSegmentedColormap


# get log probability and entropy, taking care of singularities

def CategoricalValues( N, K ):

    x_i = np.random.rand(N);

    cat_i = np.floor( K * x_i )

    return cat_i;

def BinaryValues( N ):

    x_i = np.random.rand(N);

    bin_i = (x_i < 0.5)

    return bin_i;

def OrdinalValues( N, K ):

    ord_ki = np.zeros((K,N));

    arr = np.arange(1,K+1)

    for i in range(N):
        ord_ki[:,i] = np.random.permutation(arr);

    return ord_ki;

def DiscreteValues( N, Low, High ):

    x_i = np.random.rand(N);

    disc_i = Low + np.floor( (High-Low+1) * x_i )

    return disc_i;

def ContinuousValues( N, Low, High ):

    x_i = np.random.rand(N);

    cont_i = Low + (High-Low) * x_i;

    return cont_i;



def plot_types_of_variables(Nobs=10, Kcat=5):
    """(Slide-25)"""

    # generate values of observations
    cat_i  = CategoricalValues( Nobs, Kcat );
    bin_i  = BinaryValues( Nobs );
    ord_ki = OrdinalValues( Nobs, Kcat );
    disc_i = DiscreteValues( Nobs, 0, 10 );
    cont_i = ContinuousValues( Nobs, 0, 10 );

    # generate some nice colors
    num_colors = Kcat+1
    hues = np.linspace(0, 1, num_colors)  # Evenly spaced hues
    colors = [mcolors.hsv_to_rgb([h, 1, 1]) for h in hues]

    # Plot results
    fig, axs = plt.subplots( 5,1, figsize=(8,10))

    x_i = np.arange(1,Nobs+1)

    # first subplot

    axs[0].scatter( x_i, cat_i, facecolors='yellow', edgecolor='black', s=50 )

    axs[0].set_box_aspect(1/5)
    axs[0].set_ylabel('Categorical', fontsize=14);

    # second subplot
    ix0 = np.where( bin_i==0 )
    ix1 = np.where( bin_i==1 )

    axs[1].scatter( x_i[ix0], Kcat*bin_i[ix0], facecolors='white', edgecolors='black', s=50 )
    axs[1].scatter( x_i[ix1], Kcat*bin_i[ix1], facecolors='red', edgecolors='black', s=50 )

    axs[1].set_box_aspect(1/5)
    axs[1].set_ylabel('Binary', fontsize=14);

    # third subplot
    arN = np.arange(1,Nobs+1)
    arK = np.arange(1,Kcat+1)

    x_ki = np.tile( arN, (Kcat, 1) )
    y_ki = np.tile( np.reshape( arK, (Kcat,1)), (1, Nobs) )

    for k in range(1,Kcat+1):
        ix = np.where( ord_ki==k )
        axs[2].scatter( x_ki[ix], y_ki[ix], facecolors=colors[k-1], edgecolors='black', s=50 )

    axs[2].set_box_aspect(1/5)
    axs[2].set_ylabel('Ordinal', fontsize=14);

    # fourth subplot

    axs[3].scatter( x_i, disc_i, facecolors='white', edgecolors='black', s=50 )

    for yline in range(11):
        axs[3].axhline(y=yline, color='black', linestyle='--', linewidth=0.5)

    axs[3].set_box_aspect(1/5)
    axs[3].set_ylabel('Discrete', fontsize=14);

    axs[3].set_xlabel('observation')

    # fifth subplot

    axs[4].scatter( x_i, cont_i, facecolors='red', edgecolors='black', s=50 )

    for yline in range(11):
        axs[4].axhline(y=yline, color='black', linestyle='--', linewidth=0.5)


    axs[4].set_box_aspect(1/5)
    axs[4].set_ylabel('Continuous', fontsize=14);

    axs[4].set_xlabel('observation')

    plt.tight_layout()
    plt.show()

# Creating sliders for concentrations and charge
interact(plot_types_of_variables, Nobs = (10,100,10), Ncat = (3,10,1) )
```

# **Inter-spike-interval example**

Let's turn to a neuroscience example, the distribution of inter-spike-intervals in a spike train.

Suppose we observe an irregular spike train with spike times $t_i$, where $i=1, 2, \ldots $.

We can define an inter-spike-intervals $t^\mathit{isi}_i$ by forming the difference between two successive spike times

$$t^\mathit{isi}_i = t_{i+1} - t_{i}$$

<img src="https://raw.githubusercontent.com/jochen-braun/Theoretical-Neuroscience-I/main/Lec07_Spike_train_b.png" width="500">

Possible interval durations range from $0$ to $\infty$

$$t^\mathit{isi} \in \mathbb{R}^+$$

To tabulate the outcomes, we can divide the interval duration axis into **non-overlapping** bins $\Delta t$

$\begin{eqnarray}
[0,\Delta t] \qquad [\Delta t, 2\Delta t] \qquad \ldots \qquad [i\Delta t, (i+1)\Delta t] \qquad \ldots
\end{eqnarray}$

and count the number $n_i$ of intervals observed in each bin. Obviously, when the counts $n_i$ are summed over all bins, we retrieve the ensemble size $N$ (i.e., the total number of intervals)

$$\begin{eqnarray}
\qquad\qquad \sum_i n_i = N
\end{eqnarray}$$

```python
# @title Observe irregular spike train {"vertical-output":true,"display-mode":"form"}
import matplotlib.pyplot as plt
import numpy as np
from ipywidgets import interact

# Illustrate counts, fractions, and densities
# for a distribution with linearly increasing hazard

# Rho(t) = (pi * t)/(2 * a^2);
# S(t)   = exp( -int Rho(t) dt ) = exp( - (pi * t^2)/(4 * a^2) )
# P(t)   = Rho(t) * S(t) = (pi * t)/(2 * a^2) * exp( -pi * t^2 / (4 * a^2) );

# ti = sqrt( -(4 * a^2 / pi ) * log( S(t) ) )

# int P(t) dt = 1
# int t P(t) dt = a
# int t^2 P(t) dt = 4 * a^2 / pi

def Samplelinearhazard( Nsample, m ):
    uk = np.random.rand(Nsample);
    tk = np.sqrt( -4 * m**2 * np.log( uk ) / np.pi );
    return tk

# Define an interactive plot
def plot_make_random_spiketrain(Nspike = 100):
    """(Slide-8)"""

    # parameters
    m_isi = 30;      # 30 ms

    # sample linear hazard distribution
    tk = Samplelinearhazard(Nspike, m_isi);

    ti = np.cumsum( tk );
    Ni = len( ti );

    T  = ti[Ni-1];

    # Create a figure and a grid of subplots (1 rows, 1 columns)
    fig, axs = plt.subplots(1, figsize=(9, 3))

    # First subplot
    plt.subplot(1, 1, 1)  # (nrows, ncols, plot_index)
    for i in range(0,Ni):
        plt.plot((ti[i],ti[i]), (0,1), color='black', linewidth=0.5);
    plt.title('Poisson spike train, Rate = 33 Hz', fontsize=18)
    plt.xlabel('time [ms]', fontsize=14)

    # Adjust layout to prevent overlap
    plt.tight_layout()

    # Display the plots
    plt.show()


# Creating sliders for concentrations and charge
interact(plot_make_random_spiketrain, Nspike=(100, 1000, 100))

```

# Count, fraction, and density

To report our distribution in terms independent of interval number $N$, we can convert interval **counts** into interval **fractions**

$\begin{eqnarray}
f_i = \frac{n_i}{N} = \frac{n_i}{\sum_i n_i} \qquad\qquad \sum_i f_i = 1
\end{eqnarray}$

which sum to unity. Alternatively, we can convert fractions to **percentages**, which would then sum to 100%.

To be independent of bin width $\Delta t$, we can further convert interval **fractions** into interval **densities**

$\begin{eqnarray}
p_i = \frac{f_i}{\Delta t} = \frac{n_i}{N \Delta t}, \qquad\qquad \sum_i p_i \Delta t = 1
\end{eqnarray}$

Densities **integrate** to unity, but do **not sum** to any value of interest.

When binsize becomes infinitesimally small, the probability density becomes a continuous function

$$\begin{eqnarray}
\Delta t \quad &\longrightarrow& \quad 0
\\
\\
p_i \quad &\longrightarrow& \quad p\left(t^\mathit{isi}\right)
\\
\\
\sum_i p_i \Delta t = 1 \quad &\longrightarrow& \quad \int_0^\infty p(t) \, dt = 1
\end{eqnarray}$$

```python
# @title Count, fraction, and density of inter-spike-intervals {"vertical-output":true,"display-mode":"form"}
import matplotlib.pyplot as plt
import numpy as np
from ipywidgets import interact

# Illustrate counts, fractions, and densities
# for a distribution with linearly increasing hazard

# Linear hazard     Rho(t) = (pi * t)/(2 * a^2);
# Survivor fraction S(t)   = exp( -int Rho(t) dt ) = exp( - (pi * t^2)/(4 * a^2) )
# Density           P(t)   = Rho(t) * S(t) = (pi * t)/(2 * a^2) * exp( -pi * t^2 / (4 * a^2) );

# ti = sqrt( -(4 * a^2 / pi ) * log( S(t) ) )

# int P(t) dt = 1
# int t P(t) dt = a
# int t^2 P(t) dt = 4 * a^2 / pi

def Linearhazard(  ti, m ):
    Pi = np.pi * ti / (2 * m**2) * np.exp( -np.pi * ti**2 / (4 * m**2) );
    return Pi

def Samplelinearhazard( Nsample, m ):
    uk = np.random.rand(Nsample);
    tk = np.sqrt( -4 * m**2 * np.log( uk ) / np.pi );
    return tk

def SortIntoBins( tbin, Tmin, Tmax, tlist ):

    Nmin  = round( Tmin / tbin);         # first bin
    Nmax  = round( Tmax / tbin);         # last bin
    Nbin  = Nmax - Nmin + 1;             # number of bins

    t_i   = tbin * np.linspace(Nmin, Nmax, Nbin);       # left edges
    ix    = np.digitize( tlist, t_i );

    n_i   = np.zeros_like(t_i);
    for i in range(0, Nbin):
        n_i[i] = np.sum( ix == i+1 );

    return t_i, n_i

# Define an interactive plot
def plot_count_fraction_density(binsize = 5, Nsample = 100):
    """(Slide-9)"""

    # parameters
    m_isi = 30;      # 30 ms

    T       = 100;                                       # total interbal range, in ms (15 ms)
    Nbin    = round(T/binsize);                          # number bins
    tbin_i  = binsize * np.linspace(0, Nbin, Nbin+1);    # time vector with left edges

    # get analytical density
    ti      = np.linspace(0, T, 1001);
    Pi      = Linearhazard(  ti, m_isi );

    # sample linear hazard distribution
    tk = Samplelinearhazard(Nsample, m_isi);

    # sort intervals into bins and obtain counts
    (t_i, n_i) = SortIntoBins( binsize, 0, T, tk );

    # convert counts to fractions and to densities
    f_i = n_i / Nsample;
    p_i = f_i / binsize;

    # Create a figure and a grid of subplots (2 rows, 1 columns)
    fig, axs = plt.subplots(1, 3, figsize=(9, 3))

    # First subplot
    plt.subplot(1, 3, 1)  # (nrows, ncols, plot_index)
    plt.step(t_i, n_i, where='post', color='blue')
    plt.title('Count', fontsize=18)
    plt.xlabel('time [ms]', fontsize=14)

    # Second subplot
    plt.subplot(1, 3, 2)
    plt.step(t_i, f_i, where='post', color='green')
    plt.title('Fraction', fontsize=18)
    plt.xlabel('time [ms]', fontsize=14)

    # Third subplot
    plt.subplot(1, 3, 3)
    plt.plot(ti, Pi, color='black', linestyle='--')
    plt.step(t_i, p_i, where='post', color='red')
    plt.title('Density [1/ms]', fontsize=18)
    plt.xlabel('time [ms]', fontsize=14)
    plt.text(65, 0.025, 'fraction', fontsize=10, color='red')
    plt.text(60, 0.020, 'per binwidth', fontsize=10, color='red')

    # Adjust layout to prevent overlap
    plt.tight_layout()

    # Display the plots
    plt.show()

# Creating sliders for concentrations and charge
interact(plot_count_fraction_density, binsize=(1, 10, 1), Nsample=(100, 10000, 100))

```

- Observed outcomes can be summarized in tables, histograms, or distributions.

<br>

- **Counts** tell us how many times each outcome was observed.

# **3. Central tendency, variability, and friends**

<br>

There are different ways of quantifying the **central tendency** (the 'heavy part') of a distribution.

<br>

The first one is simply the **mode**, the outcome $x = x^\mathrm{mode}$ that is most probable and observed most often.

<br>

Another useful measure is the **median**, the outcome $x = x^\mathrm{median}$ that divides our distribution in two areas of equal size, one to the right and one to the left of the median. Note that for discrete outcomes $x$, the median may not be an integer. Note further that, for skewed distributions, the median differs from the mode.

<img src="https://raw.githubusercontent.com/jochen-braun/Advanced-Statistics/main/illumedian.png" width="600">

<br>

Similarly, it can be quite useful to divide the distibution area (probability mass) into a larger number $Q$ of equal areas. These subareas are then called **quantiles**.

The quantiles for $Q=3$ are called **tertiles**, for $Q=4$ they are called **quartiles**, for $Q=5$ they are called **quintiles**, and so on.

Here is an illustration of **quartiles**:

<img src="https://raw.githubusercontent.com/jochen-braun/Advanced-Statistics/main/illuquartiles.png" width="600">

<br>

Note that there are **no analytical formulas** for medians or quantiles. You need to count your samples and divide them into subsets of equal (or almost equal) size.

The respective Matlab and Numpy functions are called *quantile()* and *numpy.quantile()*.

<br>

Last but certainly not least is the most widely used measure of central tendency, the **mean** or **expected value**, which is also the 'center of mass' in a physical sense.

(Imagine that you cut the distribution out from cardboard or plywood. If you support your piece of craftsmanship at the mean or 'center of mass', it will balance! While our distribution does not have 'mass' in the physical sence, it does have 'probability mass'!)

When we have the complete list of $N$ observations $x_i$, we compute the mean $\bar x$ as the **average value**

$$
\bar x = \frac{1}{N} \sum_{i=1}^N x_i
$$

When we have the probabilities $p_k$ of $K$ different possible observations, we can compute this as a **probabililty-weighted** average value

$$
\langle x_i \rangle = \sum_{k=1}^K p_i \cdot x_k
$$

in this case, I prefer to use the 'expectation' bracket!

<br>

In general, the mode, median and mean of a distribution take different values. Why is it, do you think, that the **mean** is larger than the **median** in the example below?

<img src="https://raw.githubusercontent.com/jochen-braun/Advanced-Statistics/main/illumeanmedianmode.png" width="600">


<br>

Of course, for many special cases the three measures are identical (e.g., all symmetric and monomodal distributions).

# Variability

By variability we mean here how 'bunched up' or 'spread out' observations or measurements are.

To quantify this, we typically use some measure of distance from the mean.  In other words, we measure how 'bunched up' or 'spread out' our observations are from the 'center of probability mass'.

<br>

One widely used measure is the **variance**, which is defined as the average squared distance from the mean or expected value $\langle x \rangle$.

When we have the complete list of $N$ observations $x_i$, we compute the variance $\mathrm{Var}(x)$ as the **average value** of the squared distance

$$
\mathrm{Var}(x) = \frac{1}{N} \sum_{i=1}^N \left(x_i - \langle x \rangle \right)^2
$$

When we have the probabilities $p_k$ of $K$ different possible observations, we can compute this as a **probabililty-weighted** average value

$$
\mathrm{Var}(x) = \sum_{k=1}^K p_i \cdot \left( x_k - \langle x \rangle \right)^2
$$

<br>

Another measured of variability is the **standard deviation**, which is defined as the square root of the variance

$$
\mathrm{Std}(x) = \sqrt{\mathrm{Var}(x)}
$$

<br>

Yet another useful measure is the **interquartile range (IQR)**, which is defined as the distance between the quantiles of $3/4$ and $1/4$. As always with quantiles, there is no analytical formula to compute this.

<img src="https://raw.githubusercontent.com/jochen-braun/Advanced-Statistics/main/illustdiqr.png" width="600">

In the illustration above, the dashed red lines indicate mean and standard deviation, whereas the colored areas indicate quartiles.

# An important equivalence

<br>

An important equivalence concerning variance is

$$
\mathrm{Var}(x) = \left\langle \left(x- \langle x \rangle \right)^2 \right\rangle = \langle x^2 \rangle - \langle x \rangle^2
$$

This may look like white magic, but is perfectly true!

<br>

**Proof:**

$$\begin{eqnarray}
\mathrm{Var}(x) = \left\langle \left(x- \langle x \rangle \right)^2 \right\rangle &=& \frac{1}{N} \sum_{i=1}^N \left(x_i - \langle x \rangle \right)^2
\\
&=& \frac{1}{N} \sum_{i=1}^N \left[ \left(x_i\right)^2 - 2 x_i \langle x \rangle + \langle x \rangle^2\right] =
\\
&=& \frac{1}{N} \sum_{i=1}^N \left(x_i\right)^2 - 2 \langle x \rangle \frac{1}{N} \sum_{i=1}^N x_i + \langle x \rangle^2 \frac{1}{N} \sum_{i=1}^N 1 =
\\
&=& \langle x^2 \rangle - 2 \langle x \rangle \langle x \rangle  + \langle x \rangle^2  =
\\
&=& \langle x^2 \rangle - \langle x \rangle^2
\end{eqnarray}$$

# **4. Properties of normal distribution**

Lastly we return to the **normal** distribution, which we have defined as a Gaussian distribution with zero mean and unit variance in its **standard form**

$$
\Phi(u) = \frac{1}{\sqrt{2\pi}} \exp\left(-\frac{u^2}{2} \right), \qquad\qquad \int_{-\infty}^{+\infty} \Phi(u) du = 1
$$

and honored with its own symbol (capital greek phi, $\Phi$).

<br>

When we transform variables $u\rightarrow x-\mu$ (with $du \rightarrow dx$) , we obtain a (**non-standard**) normal distribution with mean $\mu$ and unit variance

$$

$$

<br>

When we transform variables $u\rightarrow x/\sigma$ (with $du \rightarrow dx/\sigma$), we obtain a normal distribution with mean zero and variance $\sigma^2$

$$
P(x) dx = \frac{1}{\sqrt{2\pi}\sigma} \exp\left(-\frac{x^2}{2\sigma^2} \right)
\, dx
$$

Lastly, when we transform $u\rightarrow (x-\mu)/\sigma$ (with $du \rightarrow dx/\sigma$), we obtain a normal distribution with mean $\mu$ and variance $\sigma^2$

$$
P(x) dx = \frac{1}{\sqrt{2\pi}\sigma} \exp\left(-\frac{(x-\mu)^2}{2\sigma^2} \right)
\, dx
$$

# Mean, variance, and friends

To indicate that samples $x$ are drawn from a standard normal distribution

$$
\Phi(x) = \mathscr{N}(0,1) = \frac{1}{\sqrt{2\pi}\sigma} \exp\left(-\frac{x^2}{2} \right), \qquad\qquad \int_{-\infty}^{+\infty} \Phi(x) dx = 1
$$

we write

$$
x \sim \Phi(x) \qquad \mathrm{or} \qquad x \sim \mathscr{N}(0,1)
$$

where $\mathscr{N}()$ is an alternative notation for a normal distribution.

<br>

The mean and variance of such samples are

$$\begin{eqnarray}
\langle x \rangle &=& \int_{-\infty}^{+\infty} x \cdot \Phi(x) dx = 0
\\
\langle x^2 \rangle - \langle x \rangle^2 = \langle x^2 \rangle &=& \int_{-\infty}^{+\infty} x^2 \cdot \Phi(x) dx = 1
\end{eqnarray}$$

<br>

The standard deviation $\mathrm{Std}(x)$ is:

$$\mathrm{Std}(x) = \sqrt{\langle x^2\rangle} = 1$$

<br>

Accordingly, the skewness and kurtosis of $\Phi(x)$ are

$$
\mathrm{Skew}(x) =  \left\langle \left( \frac{ x - \langle x \rangle }{\mathrm{Std}(x)}\right)^3 \right\rangle = \langle x^3 \rangle = 0
$$

<br>

$$
\mathrm{Kurt}(x) = \left\langle \left( \frac{ x - \langle x \rangle }{\mathrm{Std}(x)}\right)^4 \right\rangle = \langle x^4 \rangle = 3
$$


As the 'normal' kurtosis has this weird value of three, statisticians have devised an additional measure called **excess kurtosis**, defined as

$$
\mathrm{ExcKurt}(x) = \mathrm{Kurt}(x) - 3.
$$

The 'excess kurtosis' of a normal distribution is zero. Now you can sleep easy.






# Proofs (optional)

To prove all this, you have to recall 'integration by parts'
$$
\int U dV = [U \cdot V] - \int V dU
$$
and the derivative of a Gaussian function
$$

= -x \cdot \exp\left(-\frac{x^2}{2}\right)$$

The rest is smooth sailing ...

 \frac{1}{\sqrt{2\pi}} \left[ -\exp\left(-\frac{x^2}{2}\right)\right]_{-\infty}^{+\infty}  = 0
\end{eqnarray}$$

$$\begin{eqnarray}

\\
&=& \frac{1}{\sqrt{2\pi}} \left[ -x \exp\left(-\frac{x^2}{2}\right) \right]_{-\infty}^{+\infty}  + \frac{1}{\sqrt{2\pi}} \int_{-\infty}^{+\infty} \exp\left(-\frac{x^2}{2}\right) dx = 1
\end{eqnarray}$$

$$\begin{eqnarray}

\end{eqnarray}$$

\end{eqnarray}$$

# Summing normally distributed samples

When we sum **two** normally distributed samples, we open a door (just a little bit), to the weird and wonderful world of **bivariate** distributions.

$$
s = x_1 + x_2, \qquad\qquad x_1 \sim \mathscr{N}(\mu_1,\sigma_1), \quad x_2 \sim \mathscr{N}(\mu_2,\sigma_2)
$$

The reason is that the value of $s$ depends on two variables that vary randomly and **independently**. In other words, it is no longer sufficient to consider different outcomes $x$ along one axis, but we must instead consider different **combinations** of outcomes along **two** axes or, equivalently, across a plane.

<br>

Our goal is to determine the mean $\langle s\rangle$ and variance $\langle s^2\rangle - \langle s\rangle^2$ of samples $s$. Of course, any normal person would simply consult the internet. However, if we are either cynical skeptics (who trust noone) or want to develop our own insight and understanding, we will integrate the joint distribution and compute the expected values.

<br>

The joint distribution (density) of independent samples $x_1$ and $x_2$ is

$$\begin{eqnarray}
P(x_1, x_2) &=& P(x_1) \cdot P(x_2) =
\\
&=&\left[\frac{1}{\sqrt{2\pi}\sigma_1}\, \exp\left(-\frac{(x_1-\mu_1)^2}{2\sigma_1^2}\right)\right]\cdot\left[\frac{1}{\sqrt{2\pi}\sigma_2} \, \exp\left(-\frac{(x_2-\mu_2)^2}{2\sigma_2^2}\right) \right] =
 \frac{1}{2\pi\sigma_1\sigma_2} \, \exp\left(-\frac{(x_1-\mu_1)^2}{2\sigma_1^2}-\frac{(x_2-\mu_2)^2}{2\sigma_2^2}\right)
\end{eqnarray}$$

This is nice to know because it makes clear what is meant by a **joint distribution**.  However, it turns out we don't even need this wonderfully explicit formula, because we can exploit the **linearity of expectations**!

<br>

The expectation for $s=x_1+x_2$ is

<br>

$$\begin{eqnarray}
\langle s \rangle
&=& \int_{-\infty}^{+\infty}\int_{-\infty}^{+\infty}  (x_1+x_2) \,  P(x_1) \, P(x_2) \, dx_1 \, dx_2 =
\\
&=& \left[\int_{-\infty}^{+\infty} x_1 \,  P(x_1) \, dx_1 \right] \left[ \int_{-\infty}^{+\infty} \, P(x_2) \, dx_2\right] + \left[\int_{-\infty}^{+\infty} \, P(x_1) \, dx_1 \right] \left[ \int_{-\infty}^{+\infty} x_2 \, P(x_2) \, dx_2\right] =
\\
&=& \langle x_1\rangle + \langle x_2\rangle
\end{eqnarray}$$



<br>

The expectation for $s^2=(x_1+x_2)^2=x_1^2 + 2 x_1 x_2 + x_2^2$ is

<br>

$$\begin{eqnarray}
\langle s^2 \rangle
&=& \int_{-\infty}^{+\infty}\int_{-\infty}^{+\infty}  (x_1^2+2x_1x_2+x_2^2) \,  P(x_1) \, P(x_2) \, dx_1 \, dx_2 =
\\
&=& \left[\int_{-\infty}^{+\infty} x_1^2 \,  P(x_1) \, dx_1 \right] \cdot 1 + 2 \left[\int_{-\infty}^{+\infty}x_1 P(x_1) \, dx_1\right] \cdot \left[ \int_{-\infty}^{+\infty} x_2 \, P(x_2) \, dx_2\right]  + 1 \cdot \left[ \int_{-\infty}^{+\infty} x_2^2 \, P(x_2) \, dx_2\right] =
\\
&=& \langle x_1^2\rangle + 2 \langle x_1\rangle \langle x_2\rangle + \langle x_2^2\rangle
\end{eqnarray}$$

<br>

Thus, the variance of $s^2=(x_1+x_2)^2$ is

$$\begin{eqnarray}
\langle s^2\rangle - \langle s\rangle^2 &=& \langle (x_1+x_2)^2\rangle - \left(\langle x_1\rangle + \langle x_2 \rangle \right)^2
\\
&=& \langle x_1^2\rangle + 2 \langle x_1\rangle \langle x_2\rangle + \langle x_2^2\rangle - \langle x_1 \rangle^2 - 2 \langle x_1\rangle \langle x_2\rangle - \langle x_2 \rangle^2 =
\\
&=& \langle x_1^2\rangle - \langle x_1 \rangle^2  + \langle x_2^2\rangle   - \langle x_2 \rangle^2 =
\\
&=& \mathrm{Var}(x_1) + \mathrm{Var}(x_2)
\end{eqnarray}$$

<br>

To summarize, when we sum independent samples from normal distributions, the result is again normally distributed! This is not true for many other distributions (with exceptions).

Moreover, we can readily identify the new normal distribution as both the **means and the variances are additive**!

In other words, the combined mean is the sum of the individual means, and the combined variance is the sum of the individual variances.

So that's nice and easy!

<br>



# Summing or multiplying expectations

In addition, our efforts have provided us with two general lessons:

**Sums of expectations:** Expectations are linear in that the expectation of a sum is simply the sum of the expectations.

$$
\langle x_1+x_2 \rangle = \langle x_1 \rangle + \langle x_2 \rangle
$$

<br>

**Products of independent expectations:** The expectation of a product is simply the product of the expectations, but only if all factors are independent!

$$
\langle x_1\cdot x_2 \rangle = \langle x_1 \rangle \cdot \langle x_2 \rangle \qquad\qquad \mathrm{iff} \quad x_{1,2} \quad \mathrm{independent}
$$

$$
\langle x_1\cdot x_2 \rangle \neq \langle x_1 \rangle \cdot \langle x_2 \rangle \qquad\qquad \mathrm{iff} \quad x_{1,2} \quad \mathrm{not \,\, independent}
$$

```python
from inspect import signature
# @title Sum of two normally distributed variables {"vertical-output":true,"display-mode":"form"}
import matplotlib.pyplot as plt
import numpy as np
from ipywidgets import interact
from scipy.stats import poisson
from scipy.stats import norm
from matplotlib.colors import LinearSegmentedColormap


def NormalDistribution( x, mu, sigma ):

    pdf_values = norm.pdf(x, loc=mu, scale=sigma)

    return pdf_values;


def plot_sum_of_normal_samples(mu1=0.0, sigma1=1.0, mu2=0.0, sigma2=1.0):
    """(Slide-25)"""

    Nres = 1000
    xlim = 5

    # range of outcomes
    x_i = np.linspace(-xlim, xlim, Nres, endpoint=True );

    # get distributions
    Pone_i = NormalDistribution( x_i, mu1, sigma1 );
    Ptwo_i = NormalDistribution( x_i, mu2, sigma2 );

    # sum and ave distributions
    Psum_i = NormalDistribution( x_i, mu1+mu2, np.sqrt(sigma1**2+sigma2**2) );

    # Plot results
    fig, axs = plt.subplots( 1,1, figsize=(8,4))

    # first subplot

    axs.plot( x_i, Pone_i, 'b', linewidth=2, label='x_1')
    axs.plot( x_i, Ptwo_i, 'r', linewidth=2, label='x_2')

    axs.fill( np.concatenate(([-xlim],x_i,[xlim])), np.concatenate(([0],Psum_i,[0])), color=[1.0,0.8,0.0], linewidth=1, edgecolor=[0,0,0], label='x_1+x_2' )

    axs.legend(loc='upper right')
    axs.set_xlabel('outcome')
    axs.set_box_aspect(1/2)
    axs.set_ylabel('density')
    axs.set_title('Two samples summed', fontsize=14);

    plt.tight_layout()
    plt.show()

# Creating sliders for concentrations and charge
interact(plot_sum_of_normal_samples, mu1 = (-4, 4, 0.5 ), sigma1 = (0.2, 4.0, 0.2), mu2 = (-4, 4, 0.5 ), sigma2 = (0.2, 4.0, 0.2) )
```

# Summing or averaging many samples

It happens quite often that we want to sum many samples from the **same** normal distribution, for example, when we compute **sample averages**!

Suppose we take $n$ samples from a normal distribution

$$
x_i \sim \mathscr{N}(\mu, \sigma), \qquad\qquad i \in \{ 1, \ldots, n\}
$$

and imagine that we repeat this many times. Imagine further that we are careful to take new samples every time, so that they will be **independent**.

How will the sample average be distributed?  Let's approach this carefully and first ask how the **sample sum** will be distributed?

<br>

We already know (from above) that the sample sum will be normally distributed

$$
s = \sum_i x_i, \qquad\qquad x \sim \mathscr{N}(\mu_\mathit{sum}, \sigma_\mathit{sum}^2)
$$

so that we need only determine the mean $\mu_\mathit{sum}$ and the variance $\sigma_\mathit{sum}^2$

<br>

We also know (again from above) that means and variances of **independent** variables add, so that we have

$$
\mu_\mathit{sum} = \sum_i \langle x_i \rangle = n \, \mu, \qquad\qquad  \sigma^2_\mathit{sum} = \sum_i \left( \langle x_i^2 \rangle - \langle x_i \rangle^2 \right) = n \, \sigma^2, \qquad\qquad \mathrm{Std}(s) = \sqrt{n} \, \sigma
$$

<br>

Very good. So now we are in a position to specify the normal distribution of the **sample average**

$$
m = \frac{1}{n} \sum_i x_i = \frac{s}{n}, \qquad\qquad m \sim \mathscr{N}(\mu_\mathit{ave}, \sigma_\mathit{ave}^2)
$$

by scaling mean and variance of the sample sum:

$$
\mu_\mathit{ave} = \frac{\mu_\mathit{sum}}{n} = \frac{n\,\mu}{n} = \mu, \qquad\qquad \sigma^2_\mathit{ave} = \frac{\sigma^2_\mathit{sum}}{n^2} = \frac{n \, \sigma^2}{n^2} = \frac{\sigma^2}{n}, \qquad\qquad \mathrm{Std}(m) = \frac{\sigma}{\sqrt{n}}
$$

<br>

We conclude that **sample averages** have the **same mean** as **individual samples** and, further

<br>

that **sample averages** have **smaller variance** than **individual samples**, specifically, the individual variance divided by sample size.

```python
from inspect import signature
# @title Average of multiple samples {"vertical-output":true,"display-mode":"form"}
import matplotlib.pyplot as plt
import numpy as np
from ipywidgets import interact
from scipy.stats import poisson
from scipy.stats import norm
from matplotlib.colors import LinearSegmentedColormap


def NormalDistribution( x, mu, sigma ):

    pdf_values = norm.pdf(x, loc=mu, scale=sigma)

    return pdf_values;


def plot_sum_of_normal_samples(mu=0.0, sigma=1.0, n=10):
    """(Slide-25)"""

    Nres = 1000
    xlim = 5

    # range of outcomes
    x_i = np.linspace(-xlim, xlim, Nres, endpoint=True );

    # get distributions
    Pone_i = NormalDistribution( x_i, mu, sigma );

    # ave distributions
    Pave_i = NormalDistribution( x_i, mu, sigma/np.sqrt(n) );

    # Plot results
    fig, axs = plt.subplots( 1,1, figsize=(8,4))

    # first subplot

    axs.plot( x_i, Pone_i, 'k', linewidth=2, label='ind')


    num = np.array(n)
    titlestring = f"{num:d}-sample average"

    axs.fill( np.concatenate(([-xlim],x_i,[xlim])), np.concatenate(([0],Pave_i,[0])), color=[1.0,0.8,0.0], linewidth=1, edgecolor=[0,0,0], label='ave' )

    axs.legend(loc='upper right')
    axs.set_xlabel('outcome')
    axs.set_box_aspect(1/2)
    axs.set_ylabel('density')
    axs.set_title(titlestring, fontsize=14);

    plt.tight_layout()
    plt.show()

# Creating sliders for concentrations and charge
interact(plot_sum_of_normal_samples, mu = (-4, 4, 0.5 ), sigma = (0.2, 4.0, 0.2), n = (1, 100, 3 ) )
```

# Exponential distribution for positive data

The exponential distribution is quite common in biology and neuroscience. For example, it describes the intervals between events that occur independently over time. This independence charaterizes many time series (e.g., spike trains, radioactive decay).


The probability density of an exponential distribution for **positive** observations is

$$
P(x) = \frac{1}{\theta} \, \exp\left(-\frac{x}{\theta} \right), \qquad\qquad \int_0^\infty P(x) dx = 1, \qquad\qquad x\geq 0
$$

where $\theta$ is a **scale** parameter.  The mean and the variance are

$$
\langle x \rangle = \theta, \qquad\qquad \langle x^2\rangle  = \theta^2 + \theta, \qquad\qquad \langle x^2\rangle - \langle x \rangle^2 = \theta
$$



```python
from inspect import signature
# @title Exponential distribution {"vertical-output":true,"display-mode":"form"}
import matplotlib.pyplot as plt
import numpy as np
from ipywidgets import interact
from scipy.stats import poisson
from scipy.stats import norm
from scipy.stats import binom


def ExponentialDistribution( x, scale ):

    pdf_values = np.exp( -x/scale) / scale;

    return pdf_values;


def plot_exponential_distribution(scale=5, N=10):
    """(Slide-25)"""

    Nres = 1000
    xlim = 3*scale

    # range of outcomes
    x = np.linspace(0, xlim, Nres, endpoint=True );

    # get distributions
    Pexp = ExponentialDistribution( x, scale );

    # sample distribution
    s_i = np.random.exponential( scale, N )


    # Plot results
    fig, axs = plt.subplots( 1, 1, figsize=(8,4))

    axs.plot( np.concatenate(([0],x)), np.concatenate(([0],Pexp)), 'r', linewidth=2, label='pdf')

    axs.plot( ([0,xlim]), ([0,0]), 'k:', linewidth=1 )

    axs.plot( s_i, np.zeros_like(s_i), 'ko', markersize=5 )

    axs.legend(loc='upper right')
    axs.set_xlim(-0.1,xlim)
    axs.set_xlabel('outcome')
    axs.set_box_aspect(1/2)
    axs.set_ylabel('density')
    axs.set_title('Exponential distribution', fontsize=14);

    plt.tight_layout()
    plt.show()

# Creating sliders for concentrations and charge
interact(plot_exponential_distribution, scale = (1,10,1), N = (10, 100, 10 )  )
```


# Gamma distribution for positive data

The Gamma distribution provides a versatile (two-parameter) family of probability distributions for **positive** observations. The exponential and chi-square distributions are special cases.

<br>

The probability density of a Gamma distribution is

$$
P(x) = \frac{1}{\Gamma(\alpha) \, x} \, (\lambda x)^\alpha \, e^{-\lambda x}\stackrel{\alpha=n}{=} \frac{1}{(n-1)! \, x} \, (\lambda x)^n \, e^{-\lambda x}, \qquad\qquad x\geq 0
$$

<br>

where $\alpha$ is a **shape** parameter and  $\lambda$ is a **rate** parameter.  Instead of a **rate** parameter, you parametrize this distribution also with a **scale** parameter $\theta = 1/\lambda$. The mean and the variance are

<br>

$$
\langle x \rangle = \frac{\alpha}{\lambda} = \lambda \, \theta, \qquad\qquad \langle x^2\rangle - \langle x \rangle^2 = \frac{\alpha}{\lambda^2} = \alpha \, \theta^2
$$

<br>

For statistial purposes (chi-square distribution), we import the gamma distribution from scipy

*from scipy.stats import gamma as gamdist*

taking care to not confuse this with

*from scipy.special import gamma as gamfunc*

The densitive PDF, cumulative CDF, and percent point function PPF are then available as

*gamdist.pdf, gamdist.cdf, gamdist.ppf*

The PDF is the "distribution height" at point $x$, the CDF the "distribution area" the the left of point $x$, and the PPF the value $x$ corresponding to ah "distribution area". Thus, the PPF is the inverse function of the CDF.




```python
from inspect import signature
# @title Gamma distribution {"vertical-output":true,"display-mode":"form"}
import matplotlib.pyplot as plt
import numpy as np
from ipywidgets import interact
from scipy.special import gamma as gammafunc
from scipy.stats import poisson
from scipy.stats import norm
from scipy.stats import binom
from scipy.stats import gamma as gammadist


def GammaDistribution( x, shape, rate ):

    pdf_values = (rate**shape) * (x**(shape-1)) * np.exp( -rate*x) / gammafunc( shape );

    return pdf_values;


def plot_gamma_distribution(shape=4, rate=0.1, N=10):
    """(Slide-25)"""

    Nres = 1000
    scale = 1 / rate
    xlim = 3*shape*scale

    # range of outcomes
    x = np.linspace(0, xlim, Nres, endpoint=True );

    # get distributions
    Pgamma = GammaDistribution( x, shape, rate );

    Palt = gammadist.pdf( x, a=shape, scale=1/rate )

    # sample distribution
    s_i = np.random.gamma(shape=shape, scale=scale, size=N)

    # Plot results
    fig, axs = plt.subplots( 1, 1, figsize=(8,4))

    axs.plot( x, Pgamma, 'r', linewidth=2, label='formula pdf')
    axs.plot( x, Palt, 'c--', linewidth=2, label='scipy pdf')

    axs.plot( ([0,xlim]), ([0,0]), 'k:', linewidth=1 )

    axs.plot( s_i, np.zeros_like(s_i), 'ko', markersize=5 )

    axs.legend(loc='upper right')
    axs.set_xlim(-0.1,xlim)
    axs.set_xlabel('outcome')
    axs.set_box_aspect(1/2)
    axs.set_ylabel('density')
    axs.set_title('Gamma distribution', fontsize=14);

    plt.tight_layout()
    plt.show()

# Creating sliders for concentrations and charge
interact(plot_gamma_distribution, shape = (1,10,1), rate = (0.1,1.0,0.1), N = (10, 100, 10 )  )
```

# Lognormal distribution for positive data

Many biological observations fall naturally on a geometrical scale, in other words, the increments are multiplicative rather than additive. For example, increments of age, brain size, or body mass compare better in multiplicative than in additive terms.

When such observations are described on a logarithmic scale -- which converts multiplicative increments into additive increments -- they are often normally distributed. In other words, we observe a skewed distribution on the linerar scale but a normal distribution on the logarithmic scale.

<br>

A requirement for considering a lognormal distribution is that the the data take positive values, $x\in\mathbb R^+$.

<br>

A lognormally distributed random variable $x$ is defined by the transformation

$$
x = e^{\mu + \sigma \, z}, \qquad\qquad x \sim \mathscr{LN}(\mu,\sigma), \qquad\qquad z \sim \mathscr{N}(0,1)
$$

where $z$ is a standard normally distributed variable.

<br>

Given the mean $\mu_x$ and variance $\sigma^2_x$ of obserations $x$, we choose

$$
\mu = \ln\left(\frac{\mu_X^2}{\sqrt{\mu_X^2 + \sigma_X^2}} \right), \qquad\qquad \sigma^2 = \ln\left(1 + \frac{\sigma_X^2}{\mu_X^2} \right)
$$


<br>

The density of a longnormal distribution is

$$
P(x) dx = \frac{1}{\sqrt{2\pi} \, x\, \sigma} \, \exp\left[-\frac{(\ln x - \mu)^2}{2 \sigma^2} \right] \, dx
$$

with mean and variance

$$
\langle x \rangle = \exp\left( \mu + \frac{\sigma^2}{2}\right), \qquad\qquad \langle x^2 \rangle - \langle x \rangle^2= \left[\exp\left(\sigma^2 \right) -1\right] \exp\left(2\mu + \sigma^2 \right)
$$

Note back-tranformation of variables $\ln x \to u$ and $\frac{dx}{x} \to du$ retrieves a normal distribution

$$
P(u) \, du = \frac{1}{\sqrt{2\pi} \, \sigma} \, \exp\left[-\frac{(u - \mu)^2}{2 \sigma^2} \right] \, du
$$

with linear mean $\mu$ and linear variance $\sigma^2$, both of which are positive.


```python
from inspect import signature
# @title Lognormal distribution {"vertical-output":true,"display-mode":"form"}
import matplotlib.pyplot as plt
import numpy as np
from ipywidgets import interact
from scipy.stats import poisson
from scipy.stats import norm
from scipy.stats import binom


def LognormalDistribution( x, mu, sigma ):

    pdf_values = np.exp( -(np.log(x) - mu)**2 / (2*sigma**2) ) / (x*np.sqrt(2*np.pi)*sigma)

    return pdf_values;

def NormalDistribution( z, mu, sigma ):

    pdf_values = np.exp( -(z - mu)**2 / (2*sigma**2) ) / (np.sqrt(2*np.pi)*sigma)

    return pdf_values;


def plot_lognormal_distribution(muX=10, sigmaX=4, N=10):
    """(Slide-25)"""

    mu = np.log( muX**2 / np.sqrt(muX**2 + sigmaX**2) )
    sigma = np.sqrt( np.log(1 + sigmaX**2/muX**2 ) );

    Meanx = np.exp(mu + sigma**2/2)
    Medx  = np.exp(mu             )
    Modx  = np.exp(mu - sigma**2  )
    Varx = (np.exp(sigma**2) - 1) * np.exp(2*mu + sigma**2)

    Nres = 1000
    xlim = Medx + 3 * np.sqrt( Varx )

    # range of outcomes
    x = np.linspace(0.1, xlim, Nres, endpoint=True );

    # get distributions
    Plnx = LognormalDistribution( x, mu, sigma );

    Plnmax = np.max(Plnx)

    z = np.linspace(mu-3*sigma, mu+3*sigma, Nres, endpoint=True )
    Pnz = NormalDistribution( z, mu, sigma );


    # sample distribution
    s_i = np.random.lognormal( mean=mu, sigma=sigma, size=N )


    # Plot results
    fig, axs = plt.subplots( 2, 1, figsize=(8,8))

    axs[0].plot( x,  Plnx, 'r', linewidth=2, label='pdf')

    axs[0].plot( ([0,xlim]), ([0,0]), 'k:', linewidth=1 )

    axs[0].plot( s_i, np.zeros_like(s_i), 'ko', markersize=5 )

    axs[0].plot( [Modx, Modx], [0,Plnmax], 'k--', linewidth=1)
    axs[0].plot( [Medx, Medx], [0,Plnmax], 'k--', linewidth=1)
    axs[0].plot( [Meanx, Meanx], [0,Plnmax], 'k--', linewidth=1)

    axs[0].legend(loc='upper right')
    axs[0].set_xlim(-0.1,xlim)
    axs[0].set_xlabel('outcome x')
    axs[0].set_box_aspect(1/2)
    axs[0].set_ylabel('density')
    axs[0].set_title('Lognormal distribution, linear axis', fontsize=14);


    axs[1].plot( z, Pnz, 'r', linewidth=2, label='pdf')

    axs[1].plot( (z[0],z[-1]), ([0,0]), 'k:', linewidth=1 )

    axs[1].plot( np.log(s_i), np.zeros_like(s_i), 'ko', markersize=5 )

    axs[1].legend(loc='upper right')
    axs[1].set_xlim((z[0],z[-1]))
    axs[1].set_xlabel('outcome ln(x)')
    axs[1].set_box_aspect(1/2)
    axs[1].set_ylabel('density')
    axs[1].set_title('Lognormal distribution, logarithmic axis', fontsize=14);

    plt.tight_layout()
    plt.show()

# Creating sliders for concentrations and charge
interact(plot_lognormal_distribution, muX = (1,10,1), sigmaX = (1,10,1), N = (10, 100, 10 )  )

```

# Bernoulli distribution

The Bernoulli distribution describes the probabilistic outcome of a single binary Bernoulli experiment. Thus is is the special case of the Binomial distribution for $n=1$.

The Bernoulli distribution is a  **discrete** **probability mass function** (pmf), with a probability $p$ of outcome $x=1$ ('success') and a probability $q=1-p$ of outcome $x=0$ ('failure').

<br>

The pmf can be written as

$$
f(x) = p \, \delta(x) + (1-p) \, \delta(n-1)
$$

where $\delta(x)$ is the Dirac function.

<br>

The mean and variance are

$$
\langle x\rangle = p, \qquad\qquad \langle x^2\rangle = p, \qquad\qquad \langle x^2\rangle - \langle x\rangle^2 = p (1-p)
$$





```python
# @title Bernoulli distribution {"vertical-output":true,"display-mode":"form"}
import matplotlib.pyplot as plt
import numpy as np
from ipywidgets import interact


# Define an interactive plot
def plot_bernoulli(p = 0.5):
    """(Slide-9)"""


    # Create a figure and a grid of subplots (1 rows, 1 columns)
    fig, axs = plt.subplots(1, 1, figsize=(6, 3))

    # first subplot
    axs.plot( [1,1], [0,p], 'r', linewidth=4, label='x=1')
    axs.plot( [0,0], [0,1-p], 'k', linewidth=4, label='x=0')

    axs.legend(loc='upper center')
    axs.set_xlim([-0.5, 1.5])
    axs.set_ylim([ 0.0, 1.0])
    axs.set_xlabel('outcome x')
    axs.set_xticks([0.0,1.0])
    axs.set_box_aspect(1/1)
    axs.set_ylabel('probability')
    axs.set_title('Bernoulli distribution', fontsize=14);


    # Adjust layout to prevent overlap
    plt.tight_layout()

    # Display the plots
    plt.show()

# Creating sliders for concentrations and charge
interact(plot_bernoulli, p=(0.0, 1.0, 0.1) )

```

# Binomial distribution for discrete events

A special and common distribution for discrete events is the binomal distribution.

The binomial distribution is the **discrete** **probability mass function** (pmf) resulting from the repetition of a binary 'Bernoulli experiment' (success/fail).

<img src="https://raw.githubusercontent.com/jochen-braun/Advanced-Statistics/main/illubinomial.png" width="600">

<br>

Probability of success $p$, probability of failure $q=1-p$, number of successes $n$, number of trials $N$.

<br>

Given probabilities $p$ and $1-p$ of success or failure in any individual trial, the pmf of $n$ successes in $N$ trials is:

$$
f(n) = {N \choose n} p^n \, (1-p)^{N-n}, \qquad\qquad \sum_{n=0}^N p(n) = 1
$$

<br>

- $p^n$ is the probability of $n$ (independent) successes.

- $(1-p)^{N-n}$ is the probability of $N-n$ (independent) failures.

- ${N \choose n} = \frac{N \, (N-1) \, \dots \, (N-n+1)}{n \, (n-1) \, \dots \, 1}$ is the number of different orderings.

<br>

# Mean and variance for Binomial distribution

$$
f(n) = {N \choose n} p^n \, (1-p)^{N-n}
$$

<br>

$$\begin{eqnarray}
\langle n \rangle &=& N \, p \qquad\qquad\qquad\qquad\qquad\textbf{mean}
\\
\\
\langle n^2\rangle &=& N (N-1) \, p^2 + N \, p  \qquad\qquad\quad\textbf{second moment}
\\
\\
\langle n^2 \rangle - \langle n\rangle^2 &=& N \, p \, (1-p) = N  p q \qquad\qquad\quad\textbf{variance }
\\
\\
\sqrt{\langle n^2 \rangle - \langle n\rangle^2} &=&  \sqrt{N  p q} \qquad\qquad\qquad\qquad\textbf{standard deviation}
\end{eqnarray}$$

<br>

For large $N\gg 1$, the Binomial distribution approaches a normal distribution with mean $\mu=N \, p$ and variance $\sigma^2 = N p \,(1-p)$




# Proofs (optional):

The proofs follows from the independence of Bernoulli experiments.  Let $x_i \in \{0,1\}$ be the outcomes of individual experiments, with

<br>

$$
\langle x_i\rangle = p, \quad \langle x_i^2\rangle = p, \quad \langle x_i^3\rangle=p \qquad\qquad \forall i  
$$

$$
\langle x_i\,x_j\rangle = p^2, \quad \langle x_i^2 \,x_j\rangle = p^2 \qquad\qquad \forall i \neq j
$$

$$
\langle x_i\,x_j\,x_k\rangle = p^3\qquad\qquad \forall i \neq j \neq k
$$

<br>

For the outcome of $N$ experiments

$$
n = x_1 + \ldots + x_N
$$

we have the first moment (mean)

$$
\langle n\rangle = \langle x_1\rangle + \ldots + \langle x_N\rangle = N \, \langle x_i\rangle = N \, p
$$

<br>

the second moment and variance

$$\begin{eqnarray}
\langle n^2\rangle &=& \left\langle \left( x_1 + \ldots + x_N \right)^2 \right\rangle =
\\
&=& N \, \langle x_i^2\rangle + N\,(N-1)\,\langle x_i\, x_j\rangle =
\\
&=& N\,p + N\,(N-1)\, p^2
\end{eqnarray}$$

<br>

$$
\left\langle n^2 - \langle n\rangle^2\right\rangle = \langle n^2 \rangle -  N^2 \, p^2  = N \, p \, (1-p) = N  p q

# Poisson distribution for discrete events

Another distribution for discrete events is the Poisson distribution (**"law of rare events"**).

Consider ever more repetitions $N$ of the binary experiment, with ever smaller probability of success, such that the expected total number of successes $\langle n \rangle$ remains the same:


<img src="https://raw.githubusercontent.com/jochen-braun/Advanced-Statistics/main/illurareevents.png" width="600">

<br>

$$
N \to \infty \qquad\qquad \lim_{N\to\infty} \langle n \rangle = \lambda
$$

<br>

In this limit, the probability of $n$ successes is given by the  **discrete Poisson distribution** ("law of rare events"):

$$
f(n) = \frac{\lambda^n}{n!} e ^{-\lambda}, \qquad\qquad \sum_{n=0}^{\infty} f(n) = 1
$$

<br>

**Computational note:**

To evaluate $f(n)$ numerically for large values of $n$, use
$$\begin{eqnarray}
f(0) &=& \exp\left[ -\lambda \right] \\
f(n) &=& \exp\left[n \ln \lambda - \lambda - \sum_{i=1}^n \ln i \right]
\end{eqnarray}$$

<br>







# Derivation from Binomial distribution (optional)

<br>

$$\begin{eqnarray}
f(n) &=&  \frac{N \cdot (N-1) \dots (N-(n-1))}{n!} \quad p^n \quad (1-p)^{N-n}=
\\
\\
\\
&=& \frac{N \cdot (N-1) \dots (N-(n-1))}{N^n} \quad \frac{(N p)^n}{n!} \quad \left(1-\frac{Np}{N}\right)^{N-n}=
\\
\\
\\
&=&\frac{\left(1-\frac{1}{N} \right) \cdot \left(1-\frac{2}{N} \right) \dots \left(1-\frac{n-1}{N} \right)}{\left(1-\frac{\lambda}{N} \right)^n} \quad \frac{\lambda^n}{n!} \quad \left(1-\frac{\lambda}{N}\right)^{N}=
\\
\\
\\
&=&  \prod_{m=0}^{n-1} \frac{1-\frac{m}{N}}{1-\frac{\lambda}{N}} \quad  \frac{\lambda^n}{n!} \quad \left(1-\frac{\lambda}{N}\right)^{N}
\end{eqnarray}$$

<br>

$$
f(n) = \prod_{m=0}^{n-1} \frac{1-\frac{m}{N}}{1-\frac{\lambda}{N}} \quad  \frac{\lambda^n}{n!} \quad \left(1-\frac{\lambda}{N}\right)^{N}\quad = \quad \frac{\lambda^n}{n!} e^{-\lambda}
$$

<br>

since with $N\to\infty$

<br>

$$
\prod_{m=0}^{n-1} \frac{1-\frac{m}{N}}{1-\frac{\lambda}{N}} = \prod_{m=0}^{n-1} \left(1 + \frac{\lambda-m}{N-\lambda}\right) \longrightarrow 1
\quad \textrm{and}  \quad \left(1-\frac{\lambda}{N}\right)^{N} \longrightarrow e^{-\lambda}
$$

<br>

QED


# Mean and variance for Poisson distribution

Poisson probability of observing $n$ events, given expectation $\lambda$:

$$
f(n) = \frac{\lambda^n}{n!} e ^{-\lambda}, \qquad\qquad \sum_{n=0}^{\infty} f(n) = 1
$$

<br>

<img src="https://raw.githubusercontent.com/jochen-braun/Advanced-Statistics/main/illupoisson.png" width="600">

<br>


<br>

$$\begin{eqnarray}
\langle n \rangle &=& \lambda \qquad\qquad\qquad\qquad\qquad\textbf{mean}
\\
\\
\langle n^2\rangle &=& \lambda^2 + \lambda  \qquad\qquad\qquad\quad\textbf{second moment}
\\
\\
\langle n^2 \rangle - \langle n\rangle^2 &=& \lambda \qquad\qquad\qquad\qquad\quad\textbf{variance }
\\
\\
\sqrt{\langle n^2 \rangle - \langle n\rangle^2} &=&  \
\sqrt{\lambda} \qquad\qquad\qquad\qquad\textbf{standard deviation}
\\
\\
\langle n^3\rangle &=& \lambda^3 + 3\lambda^2 + \lambda  \qquad\qquad\quad\textbf{third moment}
\\
\\
\left\langle \left(n - \langle n\rangle\right)^3 \right\rangle &=& \lambda \qquad\qquad\qquad\qquad\quad\textbf{third central moment}
\end{eqnarray}$$

<br>

For large $\lambda\gg 1$, the Poisson distribution approaches a normal distribution with mean $\mu=\lambda$ and variance $\sigma^2 = \lambda$

<br>
