# **1. Student's t-distribution**

We are trying to estimate the mean of $n$ samples $x_i$, which we believe to stem from a normal distribution, but unfortunately the number $n$ is rather small.

How confident can we be about our sample mean? How far from the unknown true mean can we expect it to be?

$$
m_\mathit{obs} = \frac{1}{n} \sum_i x_i
$$

To address this question, let's enter into a hypothetical world where we know everything and can do anything we want. Surely you can see the attraction of such an escapist fantasy world!

<img src="https://raw.githubusercontent.com/jochen-braun/Advanced-Statistics/main/William_Gosset.jpg" width="400">

William Gosset, who published his famous test under the pseudonym 'Student', was presumably a practical man, as he held the impressive position of "Chief Brewer" at Guinness Brewery in Dublin, Ireland.

However, he also seems to have been fond of statistical fantasy worlds. This particular excursion brought back tangible and practical benefits.  For this reason, his work was subsequently publicized and celebrated by the far more (in)famous statistician, Ronald Fisher.





# Entering the fantasy world

So, in this fantasy world, we know the underlying normal distribution of samples

$$
x\sim \mathscr{N}(\mu,\sigma)
$$

and we are able to draw many independent sets of $n$ samples,

$$
x_{ki} = \{ x_{k1}, \ldots , x_{kn}\}
$$

where $i$ indexes samples and $k$ indexes sets of samples, with sample means $m_k$ and sample variances $s^2_k$

$$
m_k = \frac{1}{n} \sum_i x_{ki}, \qquad\qquad s^2_k = \frac{1}{n} \sum_i (x_{ki}- \mu )^2,
$$


<br>

The distribution of sample means $m_k$ is

$$
m_k \sim \mathscr{N}(\mu, \sigma/\sqrt{n})
$$

but what is the distribution of sample variances $s^2_k$? To find the answer, we will need the **chi-square distribution**.


# Chi-square distribution


If $x\sim \mathscr{N}(0,1)$ is a standard normal variable, then the distribution of the square $z = x^2$ is

$$
z \sim \chi^2_1(z) = \frac{1}{\sqrt{2\pi z}} \exp\left(-\frac{z}{2}\right)
$$

which is called the chi-square distribution with one degree of freedom. (No, we won't derive this, ask Grok3, if you must know.)

<br>

The distribution of a **sum** of independent squares $z = x_1^2 + \ldots + x_n^2$

$$
z \sim \chi^2_n(z) = \frac{z^{(n/2)-1} \, e^{-z/2}}{2^{n/2} \, \Gamma\left(\frac{n}{2}\right)} \,
$$

is the chi-square distribution with $n$ degrees of freedom, which also equals a Gamma distribution with shape parameter $n/2$ and rate parameter ${1}/{2}$. To see this, compare the definition of a Gamma distribution with shape parameter $\alpha=n/2$ and rate parameter $\lambda=1/2$
$$
P(x) = \frac{1}{\Gamma(\alpha) \, x} \, (\lambda x)^\alpha \, e^{-\lambda x}
$$

<br>

So what can we say about the distribution of sample variance, $s_k^2$, now that we have figured out the distribution of a sum of $n$ independent squares, $z$?

Dividing $\chi^2$ samples $z$ by $n-1$ gives us the distribution of sample variance for a standard normal distribution, and scaling with $\sigma^2$ gives the same for a non-standard normal distribution.

$$s_k^2 = \frac{\sigma^2}{n-1} \, z, \qquad\qquad z \sim \chi^2_{n-1}$$

where $z$ follows a chi-square distribution with $n-1$ degrees (not $n$ degrees) of freedom. We use **Bessel's correction** because we had to estimate the mean before computing variance. Please use the 'toggle' in the code below to convince yourself that this improves the result noticeably, at least for small $n$!

<br>



```python
from inspect import signature
# @title Distribution of means and variances {"vertical-output":true,"display-mode":"form"}
import matplotlib.pyplot as plt
import numpy as np
from ipywidgets import interact
from scipy.stats import poisson
from scipy.stats import norm
from scipy.stats import gamma



def NormalDistribution( x, mu, sigma ):

    pdf_values = norm.pdf(x, loc=mu, scale=sigma)

    return pdf_values;

def GammaDistribution( z, n, scale ):

    pdf_values = gamma.pdf(z, a=n/2, scale=scale)

    return pdf_values;


def plot_distribution_means_variances(mu=0, sigma=1.0, n=10, toggle=0):
    """(Slide-25)"""

    Nres = 1000
    xlim = 4
    skew = 1
    Nbins = 200;
    Nsample = 10000;

    # range of outcomes
    x_i = np.linspace(-xlim, skew*xlim, Nres, endpoint=True );



    # get distributions
    Psample_i = NormalDistribution( x_i, mu, sigma );

    # sample mean and variance
    Pmean_i = NormalDistribution( x_i, mu, sigma/np.sqrt(n) );

    if toggle == 0:
        Pvariance_i = GammaDistribution( x_i, n, 2*sigma*sigma/n );
    else:
        Pvariance_i = GammaDistribution( x_i, n-1, 2*sigma*sigma/(n-1) );

    x_ik = np.random.normal( mu, sigma, (n,Nsample))

    m_k  = np.mean( x_ik, axis=0 )
    s2_k = np.var( x_ik, axis=0, ddof=1 )

    mcounts, mbins   = np.histogram( m_k, Nbins, density=True)
    s2counts, s2bins = np.histogram( s2_k, Nbins, density=True)

    mctrs = (mbins[:-1] + mbins[1:]) / 2
    s2ctrs = (s2bins[:-1] + s2bins[1:]) / 2

    # Plot results
    fig, axs = plt.subplots( 1,1, figsize=(8,4))

    # first subplot
    axs.plot( x_i, Psample_i, 'k:', linewidth=1, label='distribution of x_i')

    axs.step( mctrs, mcounts, 'r', linewidth=0.5, label='numerical m_k')
    axs.step( s2ctrs, s2counts, 'c', linewidth=0.5, label='numerical s_k^2')

    axs.plot( x_i, Pmean_i, 'k', linewidth=1, label='expected m_k')
    axs.fill( np.concatenate(([-xlim],x_i,[skew*xlim])), np.concatenate(([0],Pvariance_i,[0])), color=[1.0,0.8,0.0], linewidth=1, edgecolor=[0,0,0], label='expected s_k^2' )

    axs.legend(loc='upper right')
    axs.set_xlim([-xlim, skew*xlim])
    axs.set_xlabel('outcome')
    axs.set_box_aspect(1/2)
    axs.set_ylabel('density')
    axs.set_title('Sample mean and variance', fontsize=14);

    plt.tight_layout()
    plt.show()

# Creating sliders for concentrations and charge
interact(plot_distribution_means_variances, mu = (-4, 4, 0.5 ), sigma = (0.2, 4.0, 0.2), n=(5,100,5), toggle=(0,1,1) )
```

# Returning to the brewery


Let's now return to the brewery, where Mr. Gosset wishes to estimate state of fermentation in a very large (and valuable) tank of Guinness beer by inspecting the yeast cells in a small number of samples:

<img src="https://raw.githubusercontent.com/jochen-braun/Advanced-Statistics/main/brewers-yeast-cells.jpg" width="400">

*Brewer's yeast (approx. magnification 6250 x)*

To estimate the distribution of the observed mean from $n$ samples (how probable are different deviations from the true value?), he forms his famous **test statistic** $t$

$$
t_m = \frac{m-\mu_0}{s^2/\sqrt{n}}, \qquad m = \frac{1}{n} \sum_i x_i, \qquad s^2 = \frac{1}{n-1} \sum_i (x_i-m)^2
$$

where $\mu_0$ is a hypothetical true mean (e.g., a value to be ruled out or in).

<br>

While this is **not the z-score** of $m$ that we would obtain from the true mean $\mu_0$ and true standard deviation $\sigma_0$

$$
z_m = \frac{m-\mu_0}{\sigma_0^2/\sqrt{n}},
$$

the idea is similar in that it serves to rescale our observations such that they approximate a standard normal distribution.

<br>

Instead, the t-statistic is the ratio of two random variables, $m_k$ and $s_k^2$, that are **uncorrelated** of each other. One way of seeing this uncorrelatedness is to note that $m_k$ is a signed quantity (which depends on the sample signs, positive or negative), whereas $s_k^2$ is not (as it does not depend on sample signs).


The probability density of $t_m$ is

$$
f(t_m) = \frac{\Gamma\left(\frac{n}{2}\right)}{\sqrt{\pi (n-1)}\, \Gamma\left(\frac{n-1}{2} \right)} \, \left(1 + \frac{t_m^2}{n-1} \right)^{-\frac{n}{2}}
$$

which has heavier tails than a standard normal distribution, reflecting extra uncertainty from using $s$ instead of the true value $\sigma$.

<p><a href="https://commons.wikimedia.org/wiki/File:Student_t_pdf.svg#/media/File:Student_t_pdf.svg"><img src="https://upload.wikimedia.org/wikipedia/commons/4/41/Student_t_pdf.svg" alt="Student t pdf.svg" height="288" width="360"></a><br>By <a href="//commons.wikimedia.org/wiki/User_talk:Skbkekas" title="User talk:Skbkekas">Skbkekas</a> - <span class="int-own-work" lang="en">Own work</span>, <a href="https://creativecommons.org/licenses/by/3.0" title="Creative Commons Attribution 3.0">CC BY 3.0</a>, <a href="https://commons.wikimedia.org/w/index.php?curid=9546828">Link</a></p>

<br>

The **cumulative probabability** is defined as

$$
F(t_m) = \int_{-\infty}^{t} f(t) \, dt
$$


<p><a href="https://commons.wikimedia.org/wiki/File:Student_t_cdf.svg#/media/File:Student_t_cdf.svg"><img src="https://upload.wikimedia.org/wikipedia/commons/e/e7/Student_t_cdf.svg" alt="Student t cdf.svg" height="288" width="360"></a><br>By <a href="//commons.wikimedia.org/wiki/User_talk:Skbkekas" title="User talk:Skbkekas">Skbkekas</a> - <span class="int-own-work" lang="en">Own work</span>, <a href="https://creativecommons.org/licenses/by/3.0" title="Creative Commons Attribution 3.0">CC BY 3.0</a>, <a href="https://commons.wikimedia.org/w/index.php?curid=9546841">Link</a></p>

<br>

In Python, you can import both functions as *t.pdf* and *t.cdf* from *scipy.stats*. The inverse of the cdf (probability percentile) is available as *t.ppf*.

In Matlab, you can use the built-in functions *tpdf()* and *tcdf()*. The inverse of the cdf is called *tinv()*.

# Conclusion

<br>

- Student's t-distribution tells us how an observed sample mean is distributed around the true mean, **without knowing the true variance**.

- The distribution has heavier tails than a standard normal distribution, due to the extra uncertainty that comes from using the **sample variance**.

- The difference vanishes for large sample size, where sample variance approaches true variance.

- Given a sample size, we can use this to compute **confidence intervals** (see below).

- Given a desired level confidence, we can use it to compute the **required sample size**.

- For example, Mr. Gosset could use it to determine how many samples must be drawn from a tank of beer in order to be 99.99% certain that it meets the quality standards of Guinness!



# **3. Hypothesis testing**

In statistical "hypothesis testing", we compare an observation $x$ with a probability distribution $P(x)$ in order to determine the probability that a hypothesis is correct or incorrect.

<br>

Typical hypotheses are:

- Is the observed value **due to chance** and comparable to zero (null hypothesis $H_0$)?

- Is the observed value **not due to chance** and larger than zero (alternative hypothesis $H_1$)?

<br>

To make such determinations, we first have to choose a **level of statistical significance** $\alpha$, which sets the acceptable probability of error. To get published in a ho-hum journal, we can use $\alpha=0.05$ (or 5%), but for higher-impact journals we need $\alpha=0.01$ or even $\alpha=0.001$ (1% or 0.1%).

<br>

5% is considered 'significant', 1% 'highly significant', and 0.1% 'very highly significant'.

<br>

Next we convert our observation $x$ into a probability value $p$, using probability distribution $P(x)$ (see below for details).

<br>

Lastly, we compare our **p-value** to the chosen $\alpha$. For

$$p<=\alpha$$

we report triumphantly that our finding is ((very) highly) significant, whereas for

$$p> \alpha$$

we concede reluctantly that our finding 'does not reach significance'. We should never describe our own precious results as 'insignificant'. This odious term is reserved strictly for the results of our competitors.

<br>

By convention, we report our findings as $p<.05$, or $p<.01$, or $n.s.$ (not significant). We **do not clutter up** the manuscript by writing something like $p=0.0487$. Note further that, when reporting **p-values**, the zero before the decimal point is typically omitted: $p<.01$ rather than $p<0.01$.





# Normal distribution

In a normal distribution, $\sim$68.2% of the probability mass falls within $\pm \sigma$ of the mean,  $\sim$95.4% within $\pm 2\sigma$, and $\sim$99.8% within $\pm 3\sigma$ of the mean $\mu$.

<br>

$$
\int_{-1}^{+1}\Phi(x) \, dx \approx 0.682, \qquad
\int_{-2}^{+2}\Phi(x) \, dx \approx 0.954,\qquad
\int_{-3}^{+3}\Phi(x) \, dx \approx 0.998
$$

<br>

Thus, a significance level of $\alpha=0.05$ (or 5%) implies that our observation falls with 95% probability **within** $\pm2\sigma$ of the population mean, and with 5% probability **outside** this range.

Similarly, a significance level of $\alpha=0.001$ (or 0.1%) implies that our observation falls with 99.9% probability **within** $\pm 3\sigma$ of the population mean, and with 0.1% probability outside this larger range.

Another way of putting this that $\mu\pm 2 \sigma$ is the 'two-sigma' confidence interval and $\mu\pm 3 \sigma$ is the 'three-sigma' **confidence interval**.

<br>

All of this is intended to answer the question **how close?** (is an observation is from the distribution mean).

# P-values and Error Function Complement

If you are observing a standard normally distributed variable $z \sim \mathscr{N}(0,1)$, the p-value that corresponds to an observed (absolute) value $|z|$ is

$$
p = \mathrm{erfc}\left( \frac{|z|}{\sqrt{2}} \right) = \frac{2}{\sqrt{2\pi}} \int_{|z|}^\infty \, e^{-z^2/2} \, dz
$$

Note that this $p$-value includes the probability mass of **both tails**. For the one-sided probability, you need to divide by 2.

<br>

Both Python and Matlab provide built-in functions *erfc*

For Python you need to import

*from scipy.special import erfc*

# Cumulative normal distribution


If instead we want to answer the question **how far?** (is an observation is from the distribution mean), we use the cumulative normal distribution.

Here, $\sim$2.3% of the probability mass fall **below** $-2\sigma$ of the mean, $\sim$15.9% fall **below**  $-\sigma$, and 50% below the mean.

<br>

$$
\int_{-\infty}^{-2}\Phi(x) \, dx \approx 0.023, \qquad
\int_{-\infty}^{-1}\Phi(x) \, dx \approx 0.159,\qquad
\int_{-\infty}^{0}\Phi(x) \, dx = 0.5
$$

<br>

For example, if our observed value is $x_\mathit{obs}=-2\sigma$, it falls into the lowest 2.5% of the probability mass.  Accordingly, we can report $p<.05$ ('significant').



Similarly, if our observed value is $x_\mathit{obs}=-3\sigma$, it falls into the lowest 0.1% of the probability mass.  Accordingly, we can report $p<.001$ (very highly significant).

# Conclusion

- The drawing and reporting of statistical conclusions follows certain conventions.

- First, we choose confidence level $\alpha$ **as low as** our results allow ($0.05$ or $0.01$ or $0.001$). The lower the better!

- Next, we choose expected distribution $P(x)$ for our observations. This choice depends on the question we are asking. The expected distribution can come from a null hypothesis, from our favourite hypothesis, from another experiment, from bootstrapping, or many other sources.

- Finally, we compare our observation $x$ to the distribution $P(x)$. Does it fall near the mean or far from the mean? Specifically, does it fall **inside or outside** the confidence intervals for $P(x)$ specified by $\alpha$?

- We report the p-value of the hypothesis that we **exclude**. We can **only disprove**, never prove! This is why **small** p-values give us **high** statistical confidence.


# **4. Student's t-test for different mean**



First we consider the *Single-Sample t-Test*, which checks whether the mean $m$ of an observed sample differs significantly from a hypothetical population mean $\mu_0$.  Often, this hypothetical mean is $\mu_0=0$, so that we check whether the observed mean differs significantly from zero.

Given a sample $x_i = \{x_1, \ldots , x_n\}$ with $n$ values, we compute the mean, $m$, the corrected standard deviation, $s$, the standard error SE, and finally the t-statistic, $t$:

$$
m = \frac{1}{n}\sum_i x_i, \qquad s = \sqrt{\frac{1}{n-1}\sum_i (x_i-m)^2}, \qquad \mathrm{SE} = \frac{s}{\sqrt{n}} \qquad t = \frac{m-\mu_0}{s/\sqrt{n}} = \frac{m-\mu_0}{\mathrm{SE}}
$$

Next we compare the t-statistic to the Student's t-distribution with $n-1$ degrees of freedom, as illustrated in the code below. Specifically, we compare the observed value $t_\mathit{obs}$ to the quantiles that represent our chosen confidence bounds, for example, $q_\mathit{low}=0.025$ and $q_\mathit{high}=0.975$ for $\alpha=0.05$.

Quantiles are obtained from the inverse cumulative distribution, which is aso called the percent-point function (PPF).

In Python, you can import this as *t.ppf* from *scipy.stats*.

In Matlab, you can use the built-in functions *tinv()*.

```python
from inspect import signature
# @title Single-Sample t-Test {"vertical-output":true,"display-mode":"form"}
import matplotlib.pyplot as plt
import numpy as np
from ipywidgets import interact
from scipy.stats import norm
from scipy.stats import chi2
from scipy.stats import t


def NormalDistribution( x, mu, sigma ):

    pdf_values = norm.pdf(x, loc=mu, scale=sigma)

    return pdf_values;

def TDistribution( x, dof ):

    pdf_values = t.pdf(x, dof)

    return pdf_values;

def TQuantiles( q, dof ):

    cdf_values = t.ppf(q, dof)

    return cdf_values;

def prepare_for_fill(ix, x, Px):
    xf = np.concatenate(([x[ix[0]]], x[ix], [x[ix[-1]]]))  # Equivalent to xf = [x(ix(1)), x(ix), x(ix(end))];
    Pxf = np.concatenate(([0], Px[ix], [0]))  # Equivalent to Pxf = [0, Px(ix), 0];

    return xf, Pxf


def plot_single_sample_t_test(alpha=0.05, n=10, toggle=0):
    """(Slide-25)"""

    Nres = 1000
    xlim = 4

    Nsample = 1000;
    Nbins = 200

    mu    = 0;
    sigma = 1;

    # range of outcomes
    x_i = np.linspace(-xlim, xlim, Nres, endpoint=True );

    # distribution of individual samples
    Psample_i = NormalDistribution( x_i, mu, sigma );

    # distribution of sample mean (n samples)
    Pmean_i   = NormalDistribution( x_i, mu, sigma/np.sqrt(n) );

    # distribution of t-statistic
    Pt_i      = TDistribution( x_i, n-1 );

    # set confidence intervals and prepare their visualization
    q = [alpha/2, 1-alpha/2];

    t_low = TQuantiles( q[0], n-1 );
    t_high = TQuantiles( q[1], n-1 );

    ixlow = np.where( (x_i <= t_low) )[0];
    ixhigh = np.where( (x_i >= t_high) )[0];

    ix = np.concatenate((ixlow, ixhigh))

    (xlow, Ptlow) = prepare_for_fill(ixlow, x_i, Pt_i)
    (xhigh, Pthigh) = prepare_for_fill(ixhigh, x_i, Pt_i)

    # generate observed sample from non-standard normal distribution
    t_obs = np.random.normal( mu+2, sigma )

    # compute t statistic for random normal samples
    if toggle==1:
        x_ik = np.random.normal( mu, sigma, (n,Nsample))

        m_k  = np.mean( x_ik, axis=0 )
        s2_k = np.var( x_ik, axis=0, ddof=1 )

        t_k = (m_k - mu) / np.sqrt(s2_k/n)

        # prepare histogram for visualization

        tcounts, tbins   = np.histogram( t_k, Nbins, density=True)
        tctrs = (tbins[:-1] + tbins[1:]) / 2
    else:
        x_ik = np.random.normal( mu, sigma, (n,1))

        m_k  = np.mean( x_ik, axis=0 )
        s2_k = np.var( x_ik, axis=0, ddof=1 )

        t_k = (m_k - mu) / np.sqrt(s2_k/n)



    # Plot results
    fig, axs = plt.subplots( 1,1, figsize=(8,4))

    # normal distributions
    if toggle==1:
        axs.plot( x_i, Psample_i, 'k:', linewidth=1, label='P(x)')
    #    axs.plot( x_i, Pmean_i, 'k', linewidth=1, label='m')

    # t-distribution
    axs.fill( np.concatenate(([-xlim],x_i,[xlim])), np.concatenate(([0],Pt_i,[0])), color=[1.0,0.8,0.0], linewidth=1, edgecolor=[0,0,0], label='P(t)' )

    # tails outside the confidence range
    axs.fill( xlow, Ptlow, color=[1.0,0.0,0.0], linewidth=0.5, edgecolor=[0,0,0], label='p<alpha' )
    axs.fill( xhigh, Pthigh, color=[1.0,0.0,0.0], linewidth=0.5, edgecolor=[0,0,0] )

    # show either histogram or individual sample
    if toggle==1:
        axs.step( tctrs, tcounts, 'y', linewidth=0.5, label='numerical P(t)')
    else:
        axs.plot(t_obs, 0.01, 'b*', markersize=20, label='t_obs')

    axs.legend(loc='upper right')
    axs.set_xlim([-xlim, xlim])
    axs.set_xlabel('outcome')
    axs.set_box_aspect(1/2)
    axs.set_ylabel('density')
    axs.set_title('Distribution of t-statistic', fontsize=14);

    plt.tight_layout()
    plt.show()

# Creating sliders for concentrations and charge
interact(plot_single_sample_t_test, alpha = (0.01, 0.05, 0.01 ), n=(5,100,5), toggle=(0,1,1) )
```

# Independent (unpaired) t-test

This applies to situations in which we have two independent (unpaired) samples, for example observations $x_{1i}$ from $n_1$ wild-time mice and observations $x_{2i}$ from $n_2$ mutant mice.

Here we would like to know whether our sample means differ significantly?

<br>

The sample means are

$$
m_1 = \frac{1}{n_1} \sum_{i=1}^{n_1} x_{1i}, \qquad\qquad m_2 = \frac{1}{n_2} \sum_{i=1}^{n_2} x_{2i}
$$

and the estimated variances of **individiual samples** are

$$
s_1^2 = \frac{1}{n_1-1} \sum_{i=1}^{n_1} \left( x_{1i} - m_1\right)^2, \qquad\qquad s_2^2 = \frac{1}{n_2-1} \sum_{i=1}^{n_2} \left( x_{2i} - m_2 \right)^2
$$

and the estimated variances of **sample means** (i.e., the squared standard errors) and of the **difference between sample means** are

$$
\frac{s_1^2}{n_1}, \qquad\qquad \frac{s_2^2}{n_2}, \qquad\qquad \frac{s_1^2}{n_1} + \frac{s_2^2}{n_2}
$$

The standard error of the difference is therefore

$$
\mathrm{SE} = \sqrt{\frac{s_1^2}{n_1} + \frac{s_2^2}{n_2}}
$$

and the test statistic for the difference is

$$
t_\mathit{diff} = \frac{(m_1-m_2) - (\mu_1 - \mu_2)}{SE}
$$

For a null hypothesis $\mu_1=\mu_2$ of zero difference, the test statistic becomes

$$
t_\mathit{diff} = \frac{m_1-m_2}{\sqrt{\frac{s_1^2}{n_1} + \frac{s_2^2}{n_2}}}
$$

with $n_1+n_2-2$ degrees of freedom.



# Paired t-test

This applies to situations in which we have **paired** samples, for example observations before and after treatment/lesion of the same $n$ individuals

Again we would like to know whether our sample means differ significantly?

<br>

The average difference between sample pairs is

$$
m_D = \frac{1}{n} \sum_{i=1}^{n} d_i, \qquad\qquad d_i = x_{1i} - x_{2i}
$$

and the estimated variance of the **differences**

$$
s_D^2 = \frac{1}{n-1} \sum_{i=1}^{n} \left( d_i - m_D\right)^2
$$

The standard error of **differences** is

$$
\mathrm{SE}_D = \frac{s_D}{\sqrt{n}}
$$

and the test statistic for **differences** is

$$
t_\mathit{diff} = \frac{m_D}{s_D/\sqrt{n}}
$$

with $n-1$ degrees of freedom.



# Summary

- t-tests are useful for assessing the significance of observed sample means **when the number of samples is small**.

- For large sample numbers, the normal distribution will do.

- The **single-sample** version tests for a significant difference from a null hypothesis (e.g., zero mean).

- The **unpaired** version tests for a significant difference between the means of two independent samples.

- The **paired** version tests for a significant difference between means when the individual samples are paired (e.g., derive from the same animal before and after treatment).







# **5. F-Test for different variance**

When two distributions have similar means (i.e., the difference is not significant), they might nevertheless have different variances.

To test for differences in variance, William Gosset's approach was extended by Ronald Fisher as follows:

We have two independent samples of sizes $n_1$ and $n_2$, with variances and means

$$\begin{eqnarray}
s_1^2 = \frac{1}{n_1-1} \sum_{i=1}^{n_1} (x_{1i}-m_1)^2, \qquad &&\qquad m_1 = \frac{1}{n_1} \sum_{i=1}^{n_1} x_{1i}
\\
s_2^2 = \frac{1}{n_2-1} \sum_{i=1}^{n_2} (x_{2i}-m_2)^2, \qquad &&\qquad m_2 = \frac{1}{n_2} \sum_{i=1}^{n_2} x_{2i}
\end{eqnarray}$$

<br>

Assuming that the underlying distributions are normal, with (unknown) variances $\sigma_1^2$ and $\sigma_2^2$, we expect the sample variances (suitably rescaled) to follow $\chi^2$ distributions:

$$\begin{eqnarray}
\frac{n_1-1}{\sigma_1^2} \, s_1^2 \quad &\sim& \quad \chi^2_{n_1-1}
\\
\frac{n_2-1}{\sigma_2^2} \, s_2^2 \quad &\sim& \quad \chi^2_{n_2-1}
\end{eqnarray}$$

<br>

As Fisher was an exceptionally vain and arrogant man, it was no surprise that he called this the 'F-statistic'. The F-statistic is the ratio of the two sample variances

$$
f = \frac{s_1^2}{s_2^2}
$$

Under the null hypothesis of identical variance, $\sigma_1^2=\sigma_2^2$, we can express this in terms of $\chi^2$ samples

$$
f = \frac{\frac{(n_1-1) s_1^2}{\sigma_1^2}\frac{1}{n_1-1}}{\frac{(n_2-1)s_2^2}{\sigma_2^2}\frac{1}{n_2-1}} \quad \sim \quad \frac{\frac{\chi^2_{n_1-1}}{n_1-1}}{\frac{\chi^2_{n_2-1}}{n_2-1}}
$$

This distribution of $f$ is called an F-distribution with $(n_1-1, n_2-1)$ degrees of freedom and can be computed from a regularized incomplete beta function:

$$
F(f, d_1, d_2) = I_{\frac{f d_1}{f d_1 + d_2}} \left(\frac{d_1}{2},\frac{d_2}{2} \right)
$$

<br>

In Python, you can compute this distribution by importing *f.pdf* or *f.cdf* or *f.pmf* from stats.

In Matlab, you can use the built-in functions *fpdf( f, d1, d2)* or *fcdf( f, d1, d2 )*.

# **6. Chi-Square Test for different distributions**

Let's raise our ambition and ask a larger question: are two sample sets drawn from the same underlying distribution, or from two different distributions?

In proper statistical language: "Can we disprove, to a certain level of significance, the null hypothesis that the two data sets stem from the same underlying distribution?"

Asking whether two distributions are different is quite common in research. However, this question can arise in different contexts:
(1) Are our observations continuous or binned (in discrete categories)? (2) Are we comparing one data sample to a known reference distribution, or are we comparing data samples from two equally unknown distributions?

The accepted test for differences between **binned data sets** is the **Chi-square test**.

If your data is continuous, you can bin it, of course, but this would incur a loss of information. For this reason, continous data is typically asssessed with the **Kolmogorov-Smirnov test (KS test)**, described below.

# Chi-square comparison to a known distribution

<br>

Let's assume that our observations fall into $N$ distinct categories (i.e., they are binned) so that the can be summarized in terms of integer counts:

<br>

$$
n_i \in \mathbb{N}, \qquad i\in \{1, \dots, N\}, \qquad N_\mathit{n} = \sum_i n_i
$$

<br>

and that we have a null hypothesis about the number of observations that we expect in each bin:

<br>

$$
e_i \in \mathbb{R}, \qquad i\in \{1, \dots, N\}, \qquad N_\mathit{e} = \sum_i e_i
$$

<br>

The chi-square statistic is defined as

$$
\chi^2 = \sum_{i=1}^N \frac{(n_i-e_i)^2}{e_i}
$$

and samples a $\chi^2$-distributed with $N$ degrees of freedom

$$
\sum_i \frac{(n_i-e_i)^2}{e_i} \sim \chi^2_N
$$

assuming that the number of model events $N_e$ and observed events $N_n$ are independent.  When the model events are renormalized so that $N_e = N_n$, there are only $N-1$ degrees if freedom.

<br>

The $\chi^2$-statistic follows a Gamma distribution with shape parameter $N/2$ and scale parameter $\chi^2/2$. The cumulative distribution is an incomplete Gamma function with the same arguments.

The code below uses the scipy functions *gamma.pdf* for density, *gamma.cdf* for cumulative probability, and *gamma.ppf* for quantiles.







# Motivation of chi-square statistic


What motivates the odd form of the chi-square statistic? Under the null-hypothesis, the $n_i$ are binomially distributed

$$
n_i \sim \mathscr{B}(N_\mathit{tot}, p_i), \qquad\qquad p_i = \frac{e_i}{N_\mathit{tot}}
$$

with an expected mean and variance of

$$
\langle n_i - e_i \rangle = 0, \qquad\qquad  \left\langle (n_i - e_i)^2 \right\rangle = N_\mathit{tot} p_i \, (1-p_i) = e_i \, \left(1 - \frac{e_i}{N_\mathit{tot}}\right) \approx e_i
$$

In other words, $e_i$ approximates the variance of $n_i-e_i$, so that the square root of the fractional terms are approximately normally distributed

$$
\frac{(n_i-e_i)}{\sqrt{e_i}} \sim \mathscr{N}(0,1)
$$

and the entire sum is $\chi^2$-distributed with $N$ degrees of freedom

$$
\sum_i \frac{(n_i-e_i)^2}{e_i} \sim \chi^2_N
$$

The approximation gets better when the number of bins or the number or events per bin are large.

# Chi-square comparision of two unknown distributions

When we compare two (equally binned) data sets from unknown distributions,

$$
r_i \in \mathbb{N}, \qquad i\in \{1, \dots, N\}, \qquad N_\mathit{r} = \sum_i r_i
$$

$$
s_i \in \mathbb{N}, \qquad i\in \{1, \dots, N\}, \qquad N_\mathit{s} = \sum_i s_i
$$


the chi-square statistic becomes

$$
\chi^2 = \sum_{i=1}^N \frac{(r_i-s_i)^2}{r_i+s_i}
$$

Note that $r_i$ and $s_i$ in the denominator approximate the **variance** of $r_i$ and $s_i$, respectively. They are summed in the denominator because the variance of a difference is the sum of the individual variances.

The number of the degrees of freedom is $N$ (the number of bins), provided that $N_r$ and $N_s$ are allowed to differ.  If they are constrained to be equal, it is $N-1$.


Explanation from PTVF 2002: "A birdwatcher wants to know whether the distribution of sighted birds as a function of species is the same this year as last.  Each bin corresponds to one species. If the birdwatcher takes his data to be the first 1000 birds that he saw in each year, then the number of degrees of freedom is N-1. If he takes his data to be all the birds he saw (on specific days of both years), then the number of degrees of freedom is N. In the latter case, he is also testing whether the birds were more numerous overall in one year or the other: that is the extra degree of freedom."

# **7. Kolmogorov-Smirnov (KS) test for different distributions**

We conclude this lecture with a test that is **non-parametric** in the sense that it does not assume normally distributed observations, but that does rely on an analytical distribution. The KS test is a widely accepted tool for assessing the difference between distributions.

<br>

The KS test uses cumulative probability functions (CDFs). Suppose you have observed $N$ data points $x_i$, with $i\in\{1,2,\ldots, N\}$.  The CDF is an integer function $S(x)$ counting the fraction of data points to the left of $x$. It looks like a staircase and increments by $1/N$ at every data point $x_i$.

<br>

<img src="https://raw.githubusercontent.com/jochen-braun/Advanced-Statistics/main/illuforKStest.png" width="600">


<br>

While one can imagine many measures for the difference between CDFs, the Kolmogorov-Smirnov measure is particularly simple. It is defined as the  absolute value of the **maximal distance** between two CDFs.

<br>

When we compare the observed $S_N(x)$ in our $N$ samples to a known CDF $P(x)$, the K-S statistic is

$$
D = \max_x \left|S_N(x) - P(x) \right|
$$

whereas when comparing two observed functions $S_{N_1}(x)$ and $S_{N_2}(x)$ , the K-S statistic is

$$
D = \max_x \left|S_{N_1}(x) -  S_{N_2}(x)\right|
$$

<br>

The function for computing significance is the monotonically decreasing function

$$
Q_{KS}(\lambda) = 2\sum_{i=1}^\infty (-1)^{i-1} \, e^{-2i^2 \lambda^2}
$$

<br>

The significance level (p-value) of an observed value of D (which can disprove the null hypothesis) is given by

<br>

$$
\mathrm{Probability}(D> D_\mathit{obs}) \approx Q_{KS}\left[\left(\sqrt{N_e} + 0.12 + 0.11/\sqrt{N_e}\right)\, D_\mathit{obs}\right]
$$

<br>

where $N_e$ is the degrees of freedom. When we have one data set with $N$ samples, we have $N_e=N$. When we have two data sets with $N_1$ and $N_2$ samples, then

$$
N_e = \frac{N_1 N_2}{N_1 + N_2}
$$

<br>

Note that sensitivity of the KS-test is not independent of the location $x$ of the maximal deviation, because the distribution of deviations from $S(x)$ is not independent of $x$: large deviations are more likely in the intermediate region $S(x)\approx 0.5$ than at either extreme, $S(x)\approx 0$ or $S(x) \approx 1$.

<br>

In Python, you can import *kstest* or *ks_2samp* from scipy.stats.

In Matlab, you can use built-in function *kstest* or *kstest2*.









```python
from inspect import signature
# @title KS-test for two samples {"vertical-output":true,"display-mode":"form"}
import matplotlib.pyplot as plt
import numpy as np
from ipywidgets import interact
from scipy.stats import gamma
from scipy.stats import ks_2samp
from scipy.stats import kstwobign

def prepare_for_fill(ix, x, Px):
    xf = np.concatenate(([x[ix[0]]], x[ix], [x[ix[-1]]]))  # Equivalent to xf = [x(ix(1)), x(ix), x(ix(end))];
    Pxf = np.concatenate(([0], Px[ix], [0]))  # Equivalent to Pxf = [0, Px(ix), 0];

    return xf, Pxf

def FindMaximalDifference( x1_i, x2_i, xmin, xmax ):

    xk = np.linspace(xmin, xmax, 1000);

    S1k = np.zeros_like(xk)
    S2k = np.zeros_like(xk)

    done1 = 0;
    done2 = 0;

    len1 = len(x1_i)
    len2 = len(x2_i)

    for k in range(len(xk)):
        xnow = xk[k];
        if done1 < len1:
            if xnow > x1_i[done1]:
                done1 = done1+1;
        if done2 < len1:
            if xnow > x2_i[done2]:
                done2 = done2+ 1;


        S1k[k] = done1;
        S2k[k] = done2;

    S1k = S1k / len(x1_i);
    S2k = S2k / len(x2_i);

    return xk, S1k, S2k

def plot_ks_two_samples(alpha=0.05, N = 10, lambda1=1, lambda2=1.2, toggle=0):
    """(Slide-25)"""


    # Parameters
    N1 = N;
    N2 = N;

    # Generate gamma-distributed samples (equivalent to Poisson process interarrival times)
    x1_i = np.sort(np.random.gamma(shape=N1/2, scale=2*lambda1/N1, size=N1));
    x2_i = np.sort(np.random.gamma(shape=N2/2, scale=2*lambda2/N2, size=N2));

    # Observed cumulative distribution
    S1_i = np.arange(1, N1+1) / N1;
    S2_i = np.arange(1, N2+1) / N2;

    # Endpoints
    xmin = np.min([x1_i[0],  x2_i[0]]);
    xmax = np.max([x1_i[-1], x2_i[-1]]);

    # find maximal difference
    (xk, S1k, S2k) = FindMaximalDifference( x1_i, x2_i, xmin, xmax )

    D_obs = np.max( np.abs(S1k - S2k) )
    num = np.array(D_obs)
    D_string = f"D-ks {num:.3f}"

    ix = np.where( np.abs( S1k - S2k ) == D_obs )
    idx = ix[0][0];

    # get distribution of KS-statistic

    # Plot two distributions
    fig, axs = plt.subplots( 1, 1, figsize=(8,4))

    axs.plot( xk, S1k, 'r', linewidth=2, label='Sample 1')
    axs.plot( xk, S2k, 'b', linewidth=2, label='Sample 2')

    axs.plot( [xk[idx], xk[idx]], [S1k[idx], S2k[idx]], 'k', linewidth=1 )
    axs.plot( xk[idx], S1k[idx], 'r*', markersize=20, label='D-KS')
    axs.plot( xk[idx], S2k[idx], 'b*', markersize=20 )

    axs.text( 1.5, 0.5, D_string, fontsize=14 )

    axs.legend(loc='lower right')
    axs.set_xlim([xmin-0.1, xmax+0.1])
    axs.set_xlabel('x')
    axs.set_box_aspect(1/2)
    axs.set_ylabel('CDF')
    axs.set_title('CDFs of two samples', fontsize=14);

    plt.tight_layout()
    plt.show()

    # Perform KS test
    (D_ks, p_ks) = ks_2samp( x1_i, x2_i, alternative='two-sided', mode='asymp')
    num = np.array(p_ks)
    p_string = f"p-value {num:.3f}"

    num1 = np.array(D_ks)
    num2 = np.array(D_obs)
    D_string = f"D-ks {num1:.3f}, D-obs {num2:.3f}"

    # get distribution of KS-statistic
    Nsample = 1000;
    x = np.linspace(0, 2, Nsample, endpoint=True );


    # Probability density function (PDF)
    pdf_vals = kstwobign.pdf(x)
    pdf_vals = pdf_vals / np.max(pdf_vals)

    # Cumulative distribution function (CDF)
    cdf_vals = kstwobign.cdf(x)

    # Survivor function (SF)
    sf_vals = kstwobign.sf(x)

    # Inverse CDF (percent point function)
    quantile = kstwobign.ppf(1-alpha)  # critical value at 95% confidence

    # Manually compute p-value using kstwobign (asymptotic distribution)
    # The test statistic is scaled as sqrt(n1 * n2 / (n1 + n2)) * D
    en = np.sqrt(N1 * N2 / (N1 + N2))  # effective sample size
    scaled_stat = en * D_ks

    p_value_manual = kstwobign.sf(scaled_stat)  # survival function = 1 - CDF

    D_string = f"D-ks {scaled_stat:.3f}"

    # pad for fill
    xp  = np.concatenate(([0], x, [1]))
    sfp = np.concatenate(([0], sf_vals, [0]))

    # tails outside confidence interval
    ixhi = np.where( (quantile <= x) )[0];
    (xhi, sfhi) = prepare_for_fill(ixhi, x, sf_vals);

    # Plot results
    fig, axs = plt.subplots( 1,1, figsize=(8,4))

    axs.plot( scaled_stat, p_value_manual, 'm*', markersize=20, label='p-value' )

    axs.plot( x, pdf_vals, 'r', label='PDF')

    axs.plot( x, cdf_vals, 'b--', label='CDF')

    #axs.plot( x, sf_vals, 'm', label='CDF')

    # KS-distribution
    axs.fill( xp, sfp, color=[1.0,0.8,0.0], linewidth=1, edgecolor=[0,0,0], label='SF' )

    # tails outside the confidence range
    axs.fill( xhi, sfhi, color=[1.0,0.0,0.0], linewidth=0.5, edgecolor=[0,0,0], label='p<alpha' )

    # observed value
    #xs.plot(D_ks, p_ks, 'b*', markersize=20, label='chi2_obs')

    axs.text( 1.5, 0.5, p_string, fontsize=14 )
    axs.text( 1.5, 0.3, D_string, fontsize=14 )

    axs.legend(loc='upper left')
    axs.set_xlim([0, 2])
    axs.set_xlabel('D statistic')
    axs.set_box_aspect(1/2)
    axs.set_ylabel('p-value')
    axs.set_title('KS distributions', fontsize=14);

    plt.tight_layout()
    plt.show()

# Creating sliders for concentrations and charge

interact( plot_ks_two_samples, alpha = (0.01, 0.05, 0.01 ), N=(10,100,10), lambda1=(0.01, 2.0, 0.01), lambda2=(0.01, 2.0, 0.01), toggle=(0,1,1) )
```

```python
# @title KS-test for normality {"vertical-output":true,"display-mode":"form"}
import matplotlib.pyplot as plt
import numpy as np
from ipywidgets import interact
from scipy.stats import norm
from scipy.stats import gamma
from scipy.stats import kstest


def plot_ks_test_for_normality(a=5, N = 50, toggle=0):
    """(Slide-25)"""

    # Parameters
    fs = 18  # Font size
    xlim = 10

    # Generate gamma-distributed samples
    x_i = np.sort(np.random.gamma(shape=a, scale=5/a, size=N))

    # Observed cumulative distribution
    C_i = np.arange(1, N+1) / N;

    # Endpoints
    xmin = x_i[0]-0.1;
    xmax = x_i[-1]+0.1;

    # pad for plotting
    xp_i = np.concatenate(([xmin,x_i[0]], x_i, [x_i[-1],xmax]))
    Cp_i = np.concatenate(([0,0], C_i, [1,1]))


    # moments
    mu1    = np.mean(x_i)
    sigma1 = np.std(x_i)

    # CDF of normal distribution
    x = np.linspace(xmin, xmax, 1000)
    normCDFx = norm.cdf(x, mu1, sigma1);

    sampleCDFx = gamma.cdf(x, a=a, scale=5/a)

    # perform KS test
    D_ks, p_ks = kstest( x_i, 'norm', args=(np.mean(x_i), np.std(x_i)))

    num = np.array(p_ks)
    p_string = f"p-value {num:.3f}"

    num = np.array(D_ks)
    D_string = f"D-ks {num:.3f}"


    # Plot setup
    plt.figure(figsize=(8, 5))
    plt.plot(x, normCDFx, 'k', linewidth=1, label='Normal CDF')
    plt.plot(x, sampleCDFx, 'r', linewidth=1, label='True CDF')

    # Constructing the step function for empirical CDF
    plt.step(xp_i, Cp_i, 'r', linewidth=2, label='Empirical CDF', where='post')

    # Highlight endpoints
    plt.scatter([xp_i[0], xp_i[-1]], [Cp_i[0], Cp_i[-1]], color='r', s=100, marker='o')

    # print result of KS test
    plt.text( 0.8*xmax, 0.4, p_string, fontsize=18 )
    plt.text( 0.8*xmax, 0.1, D_string, fontsize=18 )

    # Formatting
    plt.xlabel('x', fontsize=fs)
    plt.ylabel('CDF(x)', fontsize=fs)
    plt.title('Cumulative Probability', fontsize=fs+6)
    plt.xticks([xmin, xmax], fontsize=fs)
    plt.yticks(fontsize=fs)
    plt.legend()

    # Show the plot
    plt.show()

# Run the function

interact( plot_ks_test_for_normality, a=(1,10,1), N=(50,500,50), toggle=(0,1,1) )
```
